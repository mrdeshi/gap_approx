```python
eps = 0.000000001
def near(a,b):
    if a==b:
        return True
    if a>b:
        return (a-b)<eps
    if b>a:
        return (b-a)<eps
    
```

```python
def check(arr): #true if not half-integral
    for k in arr.keys():
        #print(arr[k])
        if (not near(arr[k], 1) and not near(arr[k], 0) and not near(arr[k], 0.5)):
            return True
    return False
```

```python
import matplotlib.pyplot as plt
import numpy as np
from generator import Generator

rng = np.random.default_rng()

a, b, size = 2, 7, 10000
sample_left = rng.beta(a=a, b=b, size=size)
sample_right = rng.beta(a=b, b=a, size=size)
pop = np.concat([sample_left,sample_right])

pop*=100
pop = list(pop)
display(len(pop))
plt.hist([sample_left,sample_right],
         50, density=True, histtype='bar')
plt.show()

gen = Generator(pop)
instances=gen.gen(100)
display(len(instances))
```

```python
import tqdm
#display(instances[2])
#single_run(instances[3],100,verbose=True)
y = {}
y["t+0"] = []
y["t+1"] = []
y["t+2"] = []

for i in tqdm.tqdm(range(100)):
    best_IP_T, sol_IP, best_LP_T, sol_LP = run(instances[i],verbose=False)
    #print(best_LP_T)
    a,b,c,d = single_run(instances[i],best_LP_T,verbose=False)
    if (not b or check(b)):
        #print("FOUND")
        #print_result(b)
        continue
    for k in range(3):
        string = 't+'
        string+=str(k)
        a,b,c,d = single_run(instances[i],best_LP_T+k,verbose=False)
        s = instances[i].find_half_integral_persistence(b,column_generation=True,verbose=False)
        if(s):
            y[string].append(True)
        else:
            y[string].append(False)


    delta = best_IP_T-best_LP_T
    if (delta >0):
        print("FOUND")
        #run(instances[i],verbose=True)

display(pd.Series(y))
```
